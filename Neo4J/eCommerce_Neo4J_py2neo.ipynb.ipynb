```python
# Query 1: What is the relationship between Alice and her orders, and how are the products and categories connected to these orders?

query1 = """
MATCH (n1:Customer {name: "Alice"})-[r1:PLACED]->(n2:Order)-[r2:CONTAINS]->(n3:Product)-[r3:BELONGS_TO]->(n4:Category) 
RETURN n1,n2,n3,n4
"""

print("Query 1: What is the relationship between Alice and her orders, and how are the products and categories connected to these orders?\n")

nodes = graph.run(query1)
for node in nodes:
    print(node)
```

```python
# Query 1: What is the relationship between Alice and her orders, and how are the products and categories connected to these orders?
# In table

query1 = """
MATCH (n1:Customer {name: "Alice"})-[r1:PLACED]->(n2:Order)-[r2:CONTAINS]->(n3:Product)-[r3:BELONGS_TO]->(n4:Category) 
RETURN n1,n2,n3,n4
"""

print("Query 1: What is the relationship between Alice and her orders, and how are the products and categories connected to these orders?\n")
print(graph.run(query1).to_table())
```
